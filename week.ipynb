```python
def INSERTIONSORT(A, n):
	'''
	Sorts in non-decreasing order
	'''
	for i in range(1,n):
		j = i
		while j > 0 and A[j] > A[j-1]:
			A[j], A[j-1] = A[j-1], A[j]
			j = j - 1
```

```python
def has_duplicates(A):
    '''
    Checks if the array has duplicates
    '''
    for i in range(1,len(A)):
        for j in range(0,i):
            if A[i] == A[j]:
                return True
    return False
```

```python
def max_stones(max_weight: float, stones: list[float]) -> int:
    '''
    Returns the maximum number of stones that can be carried
    '''
    stones.sort()
    count = 0
    for i in range(n):
        if max_weight >= stones[i]:
            max_weight -= stones[i]
            count += 1
    return count
```

```python
# Define binary search that avoids using the same index twice
def binary_search(A, target, exclude_index):
    lo, hi = 0, len(A) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if A[mid] == target:
            # If found target, ensure it's not the element at exclude_index
            if mid != exclude_index:
                return True
            else:
                # Check adjacent elements in case of duplicates
                if (mid - 1 >= lo and A[mid - 1] == target) or (mid + 1 <= hi and A[mid + 1] == target):
                    return True
                return False
        elif A[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return False

def two_sum(A:list[float]) -> bool:
    '''
    Returns True if there are two numbers in the list that add up to 0
    '''
    A.sort()

    for i in range(len(A)):
        target = -A[i]
        if binary_search(A, target, i):
            return True
    # If no such pair exists return False
    return False

A = [-1, 2, 3, 4, 5, 1]
print(two_sum(A)) # True
```

```python
def two_sum(A:list[float]) -> bool:
    '''
    Returns True if there are two numbers in the list that add up to 0
    '''
    for i in range(len(A)):
        for j in range(i,len(A)):
            if A[i] + A[j] == 0:
                return True
    # If no such pair exists return False
    return False

A = [-1, 2, 3, 4, 5, 1]
print(two_sum(A)) # True
```

```python
run_time = [5,20,45,80,125]
size = [1000, 2000, 3000, 4000, 5000]

def f(run_time, A, B):
    # exponential function
    return A + B * run_time**2

param, cov = curve_fit(f, size, run_time)

# --- Create a smooth line for the regression curve ---
v0_fit = np.linspace(min(run_time), max(size), 100)
length_fit = f(v0_fit, *param)

# --- Plot the data and the regression curve ---
plt.style.use('seaborn-v0_8-darkgrid')
plt.figure(figsize=(8, 6))
plt.plot(size, run_time, 'o', label='Data')
plt.plot(v0_fit, length_fit, label='Regression curve : $y = A + Bx^2$')
plt.xlabel('Size')
plt.ylabel('Run time')
plt.legend()
plt.show()
```
